Data Cleaning

```python
df["TotalCharges"] = df["TotalCharges"].replace(" ", pd.NA)
```

```python
df["TotalCharges"] = df["TotalCharges"].astype(float)
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())
```

Analysis

Churn Rate

```python
df["Churn"].value_counts()
```

```python
ChurnPercentage = df["Churn"].value_counts(normalize=True) * 100
print(ChurnPercentage)
```

Cohort analysis by contract

```python
ContractCohort = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100
print(ContractCohort)
```

Retention trend by contract

```python
RetentionbyContract = ContractCohort["No"]
print(RetentionbyContract)
```

```python
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[0, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
)
```

Customer lifetime pattern by tenure group

```python
TenureLifetime = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100
print(TenureLifetime)
```

Monthly charges analysis by churn

```python
df.groupby("Churn")["MonthlyCharges"].mean()
```

Identify churn by internet service

```python
InternetChurn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100
print(InternetChurn)

```

Identify churn by payment method

```python
PaymentChurn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100
print(PaymentChurn)
```
